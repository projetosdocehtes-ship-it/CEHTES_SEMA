```python
import pandas as pd
import os
from inewave.nwlistop import Gtert

# Caminho base
base_path = r"C:/Users/GabrieldeSeixasMatz/Documents/Proj_Volt/NESA/ClustCMO/Procedures"

anos = [2024, 2025, 2026]
arquivos = [1, 2, 3, 4]

for ano in anos:
    dfs = []

    for i in arquivos:
        file_path = os.path.join(base_path, f"gtert00{i}_{ano}.out")
        
        try:
            d = Gtert.read(file_path)
            df = pd.DataFrame(d.valores)
            
            # média por classe e data
            df_media = (
                df
                .groupby(["classe", "data"], as_index=False)["valor"]
                .mean()
            )
            
            dfs.append(df_media)
        
        except Exception as e:
            print(f"Erro no arquivo {file_path}: {e}")

    # Junta os 4 gters do ano
    df_ano = pd.concat(dfs, ignore_index=True)
    output_path = os.path.join(base_path, f"gtert_media_{ano}.csv")
    df_ano.to_csv(output_path, index=False)

    print(f"Salvo: {output_path}")
```
